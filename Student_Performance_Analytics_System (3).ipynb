```python
import pandas as pd
import numpy as np
df=pd.read_csv('/content/student data_set.csv')
df
```

```python

print("First 5 students:")
print(df.head())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns)

print("Dataset information:")
df.info()
```

```python

print("Missing Values:")
print(df.isnull().sum())


print("Duplicate Rows:", df.duplicated().sum())

print("Duplicate Student IDs:", df["Student_ID"].duplicated().sum())

for col in ["Python_Marks", "SQL_Marks", "Maths_Marks"]:
    print(col, "Invalid Values:",
          ((df[col] < 0) | (df[col] > 100)).sum())

print("Invalid Attendance:",
      ((df["Attendance_Percentage"] < 0) |
       (df["Attendance_Percentage"] > 100)).sum())
```

```python
print("--------------------AVERAGE OF CGPA----------------------------")
AVG=df['CGPA'].mean()
print(AVG)
```

```python
print('------------------------------ASSIGNING GRADES-----------------------------=-------')
df["Grade"] = ""

for i in range(len(df)):
    if df["CGPA"][i] >= 9:
        df.loc[i, "Grade"] = "A+"
    elif df["CGPA"][i] >= 8:
        df.loc[i, "Grade"] = "A"
    elif df["CGPA"][i] >= 7:
        df.loc[i, "Grade"] = "B"
    elif df["CGPA"][i] >= 6:
        df.loc[i, "Grade"] = "C"
    elif df["CGPA"][i] >= 5:
        df.loc[i, "Grade"] = "D"
    else:
        df.loc[i, "Grade"] = "F"

(df[["Name",'Department','Python_Marks','SQL_Marks','Maths_Marks',"CGPA", "Grade"]])
```

```python
print("-------------------Python_Marks-----------------")
import numpy as np
python_marks = df["Python_Marks"]
print("Highest Python Marks:", np.max(python_marks))
print("Lowest Python Marks:", np.min(python_marks))
print("Average Python Marks:", np.mean(python_marks))
```

```python
print("-------------------SQL_Marks-----------------")
import numpy as np
SQL_Marks = df["SQL_Marks"]
print("Highest SQL Marks:", np.max(SQL_Marks))
print("Lowest SQL Marks:", np.min(SQL_Marks))
print("Average SQL Marks:", np.mean(SQL_Marks))
```

```python
print("-------------------Maths_Marks-----------------")
import numpy as np
Maths_Marks = df["Maths_Marks"]
print("Highest Maths Marks:", np.max(Maths_Marks))
print("Lowest MATHS Marks:", np.min(Maths_Marks))
print("Average Maths Marks:", np.mean(Maths_Marks))
```

```python
result = []

for i in df["CGPA"]:
    if i >= 4:
        result.append("Pass")
    else:
        result.append("Fail")

df["Result"] = result

df[["Name", "CGPA", "Result"]]
```

```python
print('----------------------------Total Marks of Each Student-------------------------------')

df["Total_Marks"] = (
    df["Python_Marks"] +
    df["SQL_Marks"] +
    df["Maths_Marks"]
)
df[["Name", "Python_Marks", "SQL_Marks",
          "Maths_Marks", "Total_Marks"]]
```

```python
print('--------------Top Performing Student-------------------------')
highest_cgpa = max(df["CGPA"])

for i in range(len(df)):
    if df.loc[i, "CGPA"] == highest_cgpa:
        print(df.loc[i, "Name"])
        print(df.loc[i, "Department"])
        print(df.loc[i, "CGPA"])
        print(df.loc[i, "Grade"])
```

```python
print('----------------------Attendance Analyse-------------------')

attendance = df["Attendance_Percentage"]

print("Highest Attendance:", np.max(attendance))
print("Lowest Attendance:", np.min(attendance))
print("Average Attendance:", np.mean(attendance))
```

```python
print('------------------STUDENT PERFORMANCE ANALYTICS SYSTEM-----------------------')
total_students = len(df)

average_cgpa = df["CGPA"].mean()

highest_cgpa = df["CGPA"].max()
lowest_cgpa = df["CGPA"].min()

passed = (df["CGPA"] >= 4).sum()
failed = (df["CGPA"] < 4).sum()

pass_percentage = (passed / total_students) * 100
fail_percentage = (failed / total_students) * 100

top_performer = df.loc[df["CGPA"].idxmax(), "Name"]

python_average = df["Python_Marks"].mean()
numpy_average = df["SQL_Marks"].mean()
pandas_average = df["Maths_Marks"].mean()

best_subject = max(
    python_average,
    numpy_average,
    pandas_average
)

if best_subject == python_average:
    best_subject = "Python"
elif best_subject == numpy_average:
    best_subject = "SQL"
else:
    best_subject = "Maths"

print("Total Students:", total_students)
print("Class Average CGPA:", round(average_cgpa, 2))
print("Highest CGPA:", highest_cgpa)
print("Lowest CGPA:", lowest_cgpa)
print("Passing Students:", passed)
print("Failing Students:", failed)
print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Fail Percentage:", round(fail_percentage, 2), "%")
print("Top Performer:", top_performer)
print("Best Subject:", best_subject)
print('------------------------PROJECT COMPLETED SUCCESSFULLY----------------')
```
